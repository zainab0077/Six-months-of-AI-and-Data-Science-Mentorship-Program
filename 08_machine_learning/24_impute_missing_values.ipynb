# Random Forest Model 

Random forests can handle missing values inherently. They can also be used to predict missing values based on the patterns learned from the other data.

Let's see how to implement random forests in Python using the Titanic dataset.

```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
from sklearn.impute import SimpleImputer

df = sns.load_dataset('titanic')

df.isnull().sum().sort_values(ascending=False)
```

```python
df.drop('deck',axis=1,inplace=True)

df.isnull().sum().sort_values(ascending=False)
```

# Encoding Categorical Variables

```python
from sklearn.preprocessing import LabelEncoder
# Define the columns to encode as a list
columns_to_encode = ['sex','embarked','who','class','embark_town','alive']
# Create a LabelEncoder dictionary to store the encoders for each column
label_encoder = {}
# Loop through the columns to encode
for col in columns_to_encode:
    # Create a LabelEncoder object for the each column
    le = LabelEncoder()
    # Fit and transform the column using the LabelEncoder
    df[col] = le.fit_transform(df[col])
    # Store the LabelEncoder object so we can use it later
    label_encoder[col] = le

df.head()
```

```python
# Create separate DataFrames for missing and non-missing values
df_with_missing = df[df['age'].isna()]
df_without_missing = df.dropna()
```

```python
print("The shape of the original dataset is: ", df.shape)
print("The shape of the dataset with missing values removed is: ", df_without_missing.shape)
print("The shape of the dataset with missing values is: ", df_with_missing.shape)
```

```python
# Display the first few rows of the dataset with missing values
df_with_missing.head()
```

```python
# Display the first few rows of the dataset without missing values
df_without_missing.head() 
```

```python
# check the names of the columns 
df.columns
```

# Train Model 

```python
# Create feature and target datasets
X = df_without_missing.drop('age', axis=1)
y = df_without_missing['age']

# Split the dataset into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Create an instance of the RandomForestRegressor class
model = RandomForestRegressor(n_estimators=40, random_state=42)

# Fit the model to the training data
model.fit(X_train, y_train)

# Make predictions on the test data
y_pred = model.predict(X_test)

# Evaluate the model's performance
print(f"Mean Absolute Error: {mean_absolute_error(y_test, y_pred)}")
print(f"Mean Squared Error: {mean_squared_error(y_test, y_pred)}")
print(f"Root Mean Squared Error: {np.sqrt(mean_squared_error(y_test, y_pred))}")
print(f"R-squared: {r2_score(y_test, y_pred)}")
```

```python
# Check for remaining missing values
df_with_missing.isnull().sum().sort_values(ascending=False)
```

# Make Prediction 

```python
# predict the missing values in the 'age' column
y_pred = model.predict(df_with_missing.drop('age', axis=1))
# Fill the missing values with the predicted values
df_with_missing['age'] = y_pred
# Check for remaining missing values
df_with_missing.isnull().sum().sort_values(ascending=False)
```

```python
# concatenate the two DataFrames 
df_complete = pd.concat([df_without_missing,df_with_missing],axis=0)

# print the shape of the new DataFrame
print("The shape of the dataset with missing values is: ", df_complete.shape)
print("The shape of the dataset without missing values is: ", df_without_missing.shape)
print("The shape of the dataset with missing values is: ", df_with_missing.shape)
```

# Decoding Categorical Variables

```python
# Decoding Categorical Variables
for col in columns_to_encode:
    # Retrieve the LabelEncoder for each column
    le = label_encoder[col]
    # Apply the inverse transform to the column in df_complete
    df_complete[col] = le.inverse_transform(df_complete[col])

df_complete.head(14)
```

```python
# print the shape of imputed DataFrame
print("The shape of the imputed dataset is: ", df_complete.shape)
```

```python
# check the missing values after imputation
df_complete.isnull().sum().sort_values(ascending=False)
```
