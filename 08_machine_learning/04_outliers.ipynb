# Handling Outliers

```python
import pandas as pd
import numpy as np
#  1.create a dataframe
df = pd.DataFrame({'Age':[7,8,9,10,11,12,13,14,34,50]})
# 2.find mean and standard deviation
mean = np.mean(df['Age'])
std = np.std(df['Age'])
# 3.find the z-scores
df['Z-Score'] = (df['Age']-mean)/std
df
```

```python
outliers = df[np.abs(df['Z-Score']) > 3]
outliers
```

```python

# print the outliers
df[np.abs(df['Z-Score']>3)]
```

```python
# print the clean data 
df[np.abs(df['Z-Score']<3)]
```

# Scipy Method 

```python
from scipy import stats
import pandas as pd
import numpy as np
#  1.create a dataframe
df = pd.DataFrame({'Age':[7,8,9,10,11,12,13,14,34,50]})
df['Z-Score'] = np.abs(stats.zscore(df['Age']))
thresold = 2.5
df[np.abs(df['Z-Score'] >= thresold)]
```

```python
df[np.abs(df['Z-Score'] < thresold)]
```

```python
from sklearn.cluster import KMeans
data = [[2,2],[3,3],[3,4],[30,30],[31,32],[32,32]]
model = KMeans(n_clusters=2,n_init=10)
model.fit(data)
labels = model.predict(data)
outlier = [data[i] for i,label in enumerate(labels) if label==1]
print(f"Data : {data}")
print(f"outlier : {outlier}")
data_clean = [data[i] for i , label in enumerate(labels) if label ==0]
print(f"Clean Data : {data_clean}")
```
