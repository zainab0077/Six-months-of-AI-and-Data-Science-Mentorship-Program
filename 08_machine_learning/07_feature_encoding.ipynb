# Feature Encoding 

Feature encoding is the process of transforming categorical features into numeric features. This is necessary because machine learning algorithms can only handle numeric features. There are many different ways to encode categorical features, and each method has its own advantages and disadvantages. In this notebook, we will explore some of the most popular methods for encoding categorical features, such as:

- Label encoding
- Ordinal encoding
- One-hot encoding
- Binary encoding

```python
# import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load dataset
df = sns.load_dataset('tips')
df.head()
```

# Label Encoding

```python
# encode time column with sklearn 
from sklearn.preprocessing import LabelEncoder , OneHotEncoder,OrdinalEncoder

le = LabelEncoder()
df['encoded_time'] = le.fit_transform(df['time'])
print(df['time'].value_counts())
print(df['encoded_time'].value_counts())
```

# Ordinal Encoding

```python
# encode the day with sklearn specific order
oe = OrdinalEncoder(categories=[['Thur','Fri','Sat','Sun']])
df['encoded_day'] = oe.fit_transform(df[['day']])
print(df['day'].value_counts())
print(df['encoded_day'].value_counts())
```

```python
#one hot encoding 
ohe = OneHotEncoder()
ohe.fit_transform(df[['sex']]).toarray()
```

# One Hot Encoding 

```python
# one hot encoding on titanic dataset 
df=sns.load_dataset('titanic')
ohe = OneHotEncoder(sparse_output=False)
encoded_embarked = ohe.fit_transform(df[['embarked']])
encoded_embarked_df = pd.DataFrame(encoded_embarked, columns=ohe.get_feature_names_out(['embarked']))
df = pd.concat([df.reset_index(drop=True), encoded_embarked_df.reset_index(drop=True)], axis=1)
```
