```python
df = sns.load_dataset('tips')

X = pd.get_dummies(df)

X_scaled = StandardScaler().fit_transform(X)
pca_3 = PCA(n_components=3)
X_pca_3 = pca_3.fit_transform(X_scaled)

from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
scatter = ax.scatter(X_pca_3[:, 0], X_pca_3[:, 1], X_pca_3[:, 2], c=df['tip'], cmap='viridis', alpha=0.8)
ax.set_xlabel('PC1')
ax.set_ylabel('PC2')
ax.set_zlabel('PC3')
ax.set_title('3D PCA on Tips Dataset')
fig.colorbar(scatter, ax=ax, label='Tip')
plt.show()
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler,LabelEncoder

df = sns.load_dataset('tips')
cat_cols = df.select_dtypes(include='category').columns


le = LabelEncoder()
for col in cat_cols:
    df[col] = le.fit_transform(df[col])


X_scaled = StandardScaler().fit_transform(df)
pca_3 = PCA(n_components=3)
X_pca_3 = pca_3.fit_transform(X_scaled)

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
scatter = ax.scatter(X_pca_3[:, 0], X_pca_3[:, 1], X_pca_3[:, 2], c=df['tip'], cmap='viridis', alpha=0.8)
ax.set_xlabel('PC1')
ax.set_ylabel('PC2')
ax.set_zlabel('PC3')
ax.set_title('3D PCA on Tips Dataset')
fig.colorbar(scatter, ax=ax, label='Tip')
plt.show()


plt.plot(np.cumsum(pca_3.explained_variance_ratio_))
plt.title('Visualize Components and their Cumulative Variance')
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.grid(True)
plt.show()

```

```python
print(f"Explained Variance Ration: {pca_3.explained_variance_ratio_}")
print(f"Cummulative Explained Variance: {np.cumsum(pca_3.explained_variance_ratio_)}")
```

```python
loadings = pd.DataFrame(
    pca_3.components_.T,
    columns=[f"PC{i+1}" for i in range(pca_3.n_components_)],
    index=df.columns
)
print("Loadings of the first three principal components:\n", loadings.head(3).round(2))

# Visualize loadings as a heatmap
sns.heatmap(loadings, annot=True, cmap="coolwarm")
plt.title("PCA Loadings Heatmap")
plt.show()
```

```python
df['PC1'] = X_pca_3[:, 0]
df['PC2'] =  X_pca_3[:, 1]
df.head()

print(f"PC Components : {pca_3.components_}")
print(f"----------------------------------------------")
print(f"PC Explained Variance : {pca_3.explained_variance_ratio_}")
print(f"----------------------------------------------")

sns.scatterplot(data=df,x='PC1',y='PC2',hue='sex')
features = pca.components_.T

for i, feature in enumerate(features):
    plt.arrow(0, 0, feature[0], feature[1], color='red', alpha=0.5,width=0.1)
    plt.text(feature[0]*1.2, feature[1]*1.2, df.columns[i], color='black', ha='center', va='center',fontsize= 12)
plt.show()
```

# PCA for Feature Extraction Before Regression

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

# Define target variable
y = df['tip']

# Split data
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)

# Apply PCA
pca = PCA(n_components=0.95)  # Keep 95% variance
X_train_pca = pca.fit_transform(X_train)
X_test_pca = pca.transform(X_test)

print(f"Original features: {X_train.shape[1]}")
print(f"Features after PCA: {X_train_pca.shape[1]}")

# Train regressor
reg = RandomForestRegressor(random_state=42)
reg.fit(X_train_pca, y_train)

# Evaluate
y_pred = reg.predict(X_test_pca)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print(f"Mean Squared Error with PCA: {mse:.4f}")
print(f"R^2 Score with PCA: {r2:.4f}")

# Compare with original features
reg_full = RandomForestRegressor(random_state=42)
reg_full.fit(X_train, y_train)
y_pred_full = reg_full.predict(X_test)
mse_full = mean_squared_error(y_test, y_pred_full)
r2_full = r2_score(y_test, y_pred_full)
print(f"Mean Squared Error without PCA: {mse_full:.4f}")
print(f"R^2 Score without PCA: {r2_full:.4f}")
```

# Kernel PCA for Nonlinear Dimensionality Reduction

```python
from sklearn.decomposition import KernelPCA
from sklearn.datasets import make_moons

# Create nonlinear data
X, y = make_moons(n_samples=500, noise=0.05, random_state=42)

# Standard PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Kernel PCA with RBF kernel
kpca = KernelPCA(n_components=2, kernel='rbf', gamma=15)
X_kpca = kpca.fit_transform(X)

# Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax1.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='viridis')
ax1.set_title('Standard PCA')
ax2.scatter(X_kpca[:, 0], X_kpca[:, 1], c=y, cmap='viridis')
ax2.set_title('Kernel PCA (RBF)')
plt.show()
```

```python
from sklearn.datasets import fetch_olivetti_faces
from sklearn.decomposition import PCA

# Load faces dataset
faces = fetch_olivetti_faces(shuffle=True, random_state=42)
X_faces = faces.data

# Apply PCA
n_components = 150  # Try different values
pca = PCA(n_components=n_components, whiten=True, random_state=42)
pca.fit(X_faces)

# Transform and inverse transform to see reconstruction
face_compressed = pca.transform(X_faces)
face_reconstructed = pca.inverse_transform(face_compressed)

# Plot original and reconstructed faces
fig, ax = plt.subplots(2, 5, figsize=(10, 4))
for i in range(5):
    ax[0, i].imshow(X_faces[i].reshape(64, 64), cmap='gray')
    ax[0, i].set_title('Original')
    ax[0, i].axis('off')
    ax[1, i].imshow(face_reconstructed[i].reshape(64, 64), cmap='gray')
    ax[1, i].set_title(f'{n_components} PCs')
    ax[1, i].axis('off')
plt.suptitle('Face Reconstruction with PCA')
plt.show()

print(f"Original dimensions: {X_faces.shape[1]}")
print(f"Compressed dimensions: {n_components}")
print(f"Variance retained: {sum(pca.explained_variance_ratio_):.2%}")
```
