# Ridge Regression 

- Definition : Ridge Regression is a type of linear regression that includes a regularization term to prevent overfitting.
- Usage : Ridge regression is used when there is multicollinearity in the data or when the number of predictors exceeds the number of observations.
- Advantages : Ridge regression can handle multicollinearity and prevent overfitting by adding a penalty term to the loss function.
- Disadvantages : Ridge regression can be less interpretable than other regression models, especially when the number of predictors is large.
- Conclusion : Ridge regression is a powerful tool for linear regression problems, particularly when dealing with multicollinearity and high-dimensional data.

# Assumptions of ridge regression: 
1. The relationship between the independent variables and the dependent variable is linear.
2. The independent variables are not perfectly multicollinear.
3. The data is not highly correlated.
4. The residuals of the model are normally distributed.
5. The variance of the residuals is constant (homoscedasticity).
6. The model is correctly specified, meaning that all relevant variables are included and irrelevant variables are excluded.
7. The model is not overfitting the data, meaning that the model is not too complex and the data is not too simple.
8. The model is not underfitting the data, meaning that the model is not too simple and the data is not too complex.
9. The model is not suffering from multicollinearity, meaning that the independent variables are not perfectly correlated.
10. The model is not suffering from heteroscedasticity, meaning that the variance of the residuals is constant across all levels of the independent variables.
11. The model is not suffering from autocorrelation, meaning that the residuals are not correlated with each other.

# Ridge Regression VS Simple Linear Regression 

| Feature                      | Simple Linear Regression                         | Ridge Regression                                      |
|-----------------------------|---------------------------------------------------|--------------------------------------------------------|
| **Definition**              | A method to model the relationship between a dependent variable and one independent variable. | A regularized version of linear regression that includes L2 penalty to reduce overfitting. |
| **Equation**                | \( y = \beta_0 + \beta_1 x + \epsilon\)         | \( y = \beta_0 + \beta_1 x + \lambda \sum \beta_j^2 + \epsilon\) |
| **Penalty Term**            | None                                              | L2 penalty (sum of squares of coefficients)            |
| **Purpose of Penalty**      | Not applicable                                    | Shrinks coefficients to prevent overfitting            |
| **Overfitting Handling**    | More prone to overfitting                        | Less prone to overfitting                              |
| **Bias-Variance Tradeoff**  | Lower bias, higher variance                      | Higher bias, lower variance                            |
| **Feature Selection**       | Does not perform feature selection               | Does not set coefficients to zero                      |
| **When to Use**             | When data is simple and multicollinearity is not an issue | When data has multicollinearity or needs regularization |
| **Interpretability**        | Easy to interpret                                | Less interpretable due to regularization               |
| **Computational Cost**      | Lower                                             | Slightly higher due to additional penalty term         |


```python
from sklearn.linear_model import Ridge
import numpy as np

# CREATE EXAMPLE DATA 
X = np.array([[1,1],
              [1,2],
              [2,2],
              [2,3]])
y = np.dot(X, np.array([1, 2])) + 3
# Ridge regression model 
ridge = Ridge(alpha=1.0) # alpha is the regularization strength (lambda in the equation)
# Fit the model
ridge.fit(X, y)

# Print the coefficients and intercept
print(f"Coefficient :{ridge.coef_}, Intercept :{ridge.intercept_}") # 
```

- Coeifficient is the slope of the line, intercept is the y-intercept that  value of the line at x=0  and it is the point where the line crosses the y-axis.

```python
import seaborn as sns
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error,mean_absolute_percentage_error
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import numpy as np

# Load Titanic dataset
df = sns.load_dataset('titanic')
df.head()
```

## **Step 2 Preprocessing**

```python
# Selecting a subset of columns for simplicity
columns_to_use = ['survived', 'pclass', 'sex', 'age', 'fare']
df = df[columns_to_use]

# Handling missing values
df['age'].fillna(df['age'].median(), inplace=True)

# Define feature and target variable
X = df.drop('survived', axis=1)
y = df['survived']

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

## **Step 3 Creating a pipeline**

```python
# define a pipeline for OHE and model 
categorical_features = ['sex']
numerical_features = ['pclass','age','fare']

# Preprocessor 
preprocessor = ColumnTransformer(
    transformers=[
        ('num', 'passthrough', numerical_features),
        ('cat', OneHotEncoder(), categorical_features)
    ])
# Linear Regression Pipeline
lr_pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                           ('model', LinearRegression())])  # Using Linear Regression

# Ridge Regression Pipeline
ridge_pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                           ('model', Ridge(alpha=1.0))])  # Using Ridge regression

```

## **Step 4 Train and Evaluation**

```python
# train and evaluate linear regression 
lr_pipeline.fit(X_train,y_train)
lr_pred = lr_pipeline.predict(X_test)
lr_mse = mean_squared_error(y_test,lr_pred)
lr_mae = mean_absolute_error(y_test,lr_pred)
lr_r2 = r2_score(y_test,lr_pred)
lr_mape = mean_absolute_percentage_error(y_test,lr_pred)

print(f"Linear Regression MSE : {lr_mse}")
print(f"Linear Regression MAE : {lr_mae}")
print(f"Linear Regression R2 : {lr_r2}")
print(f"Linear Regression MAPE : {lr_mape}")

# train and evaluate ridge regression 
ridge_pipeline.fit(X_train,y_train)
ridge_pred = ridge_pipeline.predict(X_test)
ridge_mse = mean_squared_error(y_test,ridge_pred)
ridge_mae = mean_absolute_error(y_test,ridge_pred)
ridge_r2 = r2_score(y_test,ridge_pred)
ridge_mape = mean_absolute_percentage_error(y_test,ridge_pred)
print("-"*50)
print("-"*50)
print(f"Ridge Regression MSE : {ridge_mse}")
print(f"Ridge Regression MAE : {ridge_mae}")
print(f"Ridge Regression R2 : {ridge_r2}")
print(f"Ridge Regression MAPE : {ridge_mape}")
```
