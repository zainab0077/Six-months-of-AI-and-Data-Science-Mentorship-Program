# Data Inconsistencies 

## 1.Inconsistent Formats 

```python
# import libraries for ml tasks 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
data = {
    'date': ['2021-12-01', '01-12-2022', '2022/12/01', '12-01-2021'],
    'country': ['USA', 'U.S.A.', 'America', 'United States'],
    'name': ['Arsal', 'Ahmed', 'Abdullah', 'Arsal'],
    'sales_2020': [100, 200, None, 200],
    'sales_2021': [None, 150, 300, 150]
}
# make pandas dataframe
df = pd.DataFrame(data)
df
```

### 1.Convert to Datetime: 
Use pd.to_datetime with errors='coerce' to convert the column. This will set invalid dates to NaT.

```python
# standardizing the date column 
df['date']=pd.to_datetime(df['date'],errors='coerce')
df['date']=df['date'].dt.strftime('%y-%m-%d')
df
```

### 2.Fill Missing Values: 
Use methods like fillna() to impute missing values. You can fill with a specific date, the mean date, or forward/backward fill.

```python
# Step 1: Convert to datetime, coercing errors
df['date'] = pd.to_datetime(df['date'], errors='coerce')

# Step 2: Check for NaT values
print("Before filling NaT values:")
print(df)

# Step 3: Impute or fill missing values
# Example: Fill NaT with a specific date (e.g., '2021-01-01')
df['date'].fillna(pd.Timestamp('2001-12-21'), inplace=True)

# Alternatively, use forward fill
# df['date'].fillna(method='ffill', inplace=True)

# Print the DataFrame after imputation
print("\nAfter filling NaT values:")
print(df)
```

> Interpretation:
1. The pd.to_datetime function converts the date column, coercing invalid entries to NaT.
2. You can inspect the DataFrame to see which values are NaT.
3. Use fillna() to replace NaT values. You can choose to fill with a specific date, or apply methods like forward fill (ffill) or backward fill (bfill) based on your needs.


This approach ensures that your date column is consistent and ready for further analysis.

## 2.Naming Convention

```python
df.head()
```

```python
# Harmonize the Name of Country 
mapping = {'USA': 'United States', 'U.S.A.': 'United States', 'America': 'United States'}
df['country'] = df['country'].replace(mapping)
df
```

## 3.Typographical Inconsistency

```python
# correct the typographical inconsistency in the name column
df['name']=df['name'].replace({'Arsal':'Arsalan'})
df
```

## 4.Remove Duplicates 

```python
df.drop_duplicates(subset='name',inplace=True)
df
```
