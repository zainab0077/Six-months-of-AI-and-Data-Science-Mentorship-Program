# Model Notebook for Coding in SK-LEARN

```python
 # import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('tips')
df.head()
```

# Linear Regresssion 

```python
# Import necessary libraries
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Initialize the StandardScaler to standardize features
scaler = StandardScaler()

# Scale the 'total_bill' feature to have a mean of 0 and a standard deviation of 1
df['total_bill'] = scaler.fit_transform(df[['total_bill']])

# Define the feature set (X) and target variable (y)
X = df[['total_bill']]  # Features: total bill amount
y = df['tip']           # Target: tip amount

# Split the dataset into training and testing sets
# 80% of the data will be used for training, and 20% for testing
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=42)

# Initialize the Linear Regression model
model = LinearRegression()

# Fit the model on the training data
model.fit(X_train, y_train)

# Make predictions on the testing set
y_pred = model.predict(X_test)

# Note: Consider evaluating the model's performance using metrics such as Mean Absolute Error (MAE) or R-squared.
```

 Interpretation:
 
> StandardScaler: 
Used to standardize features to improve model performance.

> train_test_split: 
Splits the dataset for training and testing to validate model performance.

> LinearRegression: 
A basic linear regression model to predict tips based on total bills.

> Random State: 
Ensures reproducibility of the results.

```python
y_pred = model.predict([[50]])
# Assuming y_pred contains the scaled predictions
# Reshape y_pred to 2D array for inverse_transform
y_pred_scaled = y_pred.reshape(-1, 1)

# Inverse transform the predicted values to get original scale
y_pred_original = scaler.inverse_transform(y_pred_scaled)

# Print the descaled predictions
print(y_pred_original)
```

# Performance Metrics 

```python
# Import necessary metrics
from sklearn.metrics import mean_absolute_error, r2_score

# Calculate Mean Absolute Error (MAE)
mae = mean_absolute_error(y_test, y_pred)

# Calculate R-squared (R²) score
r_squared = r2_score(y_test, y_pred)

# Print MAE with conditional statements
if mae < 1:
    print(f'Mean Absolute Error (MAE): {mae:.2f} - Excellent')
elif mae < 5:
    print(f'Mean Absolute Error (MAE): {mae:.2f} - Good')
else:
    print(f'Mean Absolute Error (MAE): {mae:.2f} - Needs Improvement')

# Print R-squared with conditional statements
if r_squared >= 0.8:
    print(f'R-squared (R²): {r_squared:.2f} - Excellent Fit')
elif r_squared >= 0.5:
    print(f'R-squared (R²): {r_squared:.2f} - Fair Fit')
else:
    print(f'R-squared (R²): {r_squared:.2f} - Poor Fit')
```
