# **Inverse encoding/ Inverse transform /Decoding categorical variables**

Inverse encoding is the process of converting the encoded data back to its original form. This is done by using the **inverse_transform()** method. This method is available for the following encoders:

- LabelEncoder
- OrdinalEncoder
- OneHotEncoder

```python
# import libraries
import pandas as pd
import numpy as numpy
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, OrdinalEncoder
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
# load the dataset
df = sns.load_dataset('titanic')
# display the first few rows of the dataset
df.head()
```

```python
# impute missing values 
df['age'].fillna(df['age'].mean(), inplace=True)
df['embark_town'].fillna(df['embark_town'].mode()[0], inplace=True)
df['embarked'].fillna(df['embarked'].mode()[0], inplace=True)

# drop columns that are not needed
df.drop('deck',axis=1,inplace=True)
```

# Label Encoder

```python
# encode all the object & categorical variables 
le_sex = LabelEncoder()
le_embarked = LabelEncoder()
le_embark_town = LabelEncoder()
le_who = LabelEncoder()
le_alive = LabelEncoder()
le_alone = LabelEncoder()

df['sex'] = le_sex.fit_transform(df['sex'])
df['embarked'] = le_embarked.fit_transform(df['embarked'])
df['embark_town'] = le_embark_town.fit_transform(df['embark_town'])
df['who'] = le_who.fit_transform(df['who'])
df['alive'] = le_alive.fit_transform(df['alive'])
df['alone'] = le_alone.fit_transform(df['alone'])

df.head()
```

```python
# Apply Inverse Transform to decode the encoded variables
df['sex'] = le_sex.inverse_transform(df['sex'])
df.head()
```

# OHE

```python
ohe_sex = OneHotEncoder(sparse_output=False)
ohe_embarked = OneHotEncoder(sparse_output=False)
ohe_embark_town = OneHotEncoder(sparse_output=False)
ohe_who = OneHotEncoder(sparse_output=False)
ohe_alive = OneHotEncoder(sparse_output=False)
ohe_alone = OneHotEncoder(sparse_output=False)

# One-hot encode the categorical variables
df['sex'] = ohe_sex.fit_transform(df[['sex']])
df['embarked'] = ohe_embarked.fit_transform(df[['embarked']])
df['embark_town'] = ohe_embark_town.fit_transform(df[['embark_town']])
df['who'] = ohe_who.fit_transform(df[['who']])
df['alive'] = ohe_alive.fit_transform(df[['alive']])
df['alone'] = ohe_alone.fit_transform(df[['alone']])

df.head()
```
