## Transform Size Column

- Converting size from bytes to MB and creating new column with name "Size_MB"

```python
df['Size_MB'] = df['Size_Bytes'].apply(lambda x:(x/1024)/1024)
```

```python
df['Size_GB'] = df['Size_MB'] / 1024
```

```python
df.drop(columns='Size_MB',axis=1,inplace=True)
```

# Handling Missing Values

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
# Percentage OF Missing Values in each column of the dataset
(df.isnull().sum()*100/len(df)).sort_values(ascending=False)
```

---
## Size Column 

```python
print(f"Total Missing Values in the Size Column : {df['Size_Bytes'].isnull().sum()}")
```

- There are 224 rows that have not mentioned values for application size

```python
df['Size_Bytes'].fillna(df['Size_Bytes'].mean(),inplace=True)
df['Size_MB'].fillna(df['Size_MB'].mean(),inplace=True)
```

- We replaced the missing values with mean of the application size

```python
print(f"Missing Values in  Size Column After Imputing : {df['Size_Bytes'].isnull().sum()}")
```

```python
df.drop(columns='Size_Bytes',axis=1,inplace=True)
```

- So, the size has been replaced with mean of the application size

---
Lets check the price column for imputation 
## Price Column

```python
print(f"Mode of Price : {df['Price'].mode()[0]}")
```

```python
df['Price'].fillna(df['Price'].mode()[0],inplace=True)
```

```python
df['Price'].isnull().sum()
```

- So , null values of price column has been filled with frequently occuring values of price column
