# Parallel Coordinate Plot for K-Means Clustering

A parallel coordinate plot is a visualization technique for multidimensional data where each variable is represented by a vertical axis, and data points are shown as lines connecting their values across these axes. It's particularly useful for visualizing clusters in high-dimensional data.


## Key Components Explained:

1. **Data Generation**: We use `make_blobs` to create synthetic data with 5 features and 3 natural clusters.

2. **K-Means Clustering**: The algorithm groups the data points into 3 clusters.

3. **Parallel Coordinates Plot**: 
   - Each vertical axis represents one feature
   - Each line represents one data point
   - Colors represent cluster assignments
   - You can see how different clusters have different patterns across the features

## Customization Options:

- Change the number of clusters by modifying `n_clusters`
- Adjust the number of features with `n_features`
- Use different color scales (e.g., `px.colors.sequential.Viridis`)
- Add more samples by changing `n_samples`
- For real-world data, replace the synthetic data with your dataset

This visualization helps you understand:
- Which features are most important for cluster separation
- How clusters differ across dimensions
- Whether your clustering results make sense

## Creating a Parallel Coordinate Plot for K-Means with Plotly

```python
import numpy as np
import pandas as pd
import plotly.express as px
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

# 1. Generate sample data
n_samples = 200
n_features = 5
n_clusters = 3

# Create synthetic data with make_blobs
X, y = make_blobs(n_samples=n_samples, 
                  n_features=n_features, 
                  centers=n_clusters,
                  cluster_std=1.5,
                  random_state=42)

# Convert to DataFrame for easier handling
feature_columns = [f'Feature_{i}' for i in range(n_features)]
df = pd.DataFrame(X, columns=feature_columns)

# 2. Perform K-Means clustering
kmeans = KMeans(n_clusters=n_clusters, random_state=42)
clusters = kmeans.fit_predict(X)
df['Cluster'] = clusters.astype(str)  # Convert to string for categorical coloring

# 3. Create parallel coordinates plot
fig = px.parallel_coordinates(
    df,
    color='Cluster',  # Color by cluster assignment
    dimensions=feature_columns,
    color_continuous_scale=px.colors.qualitative.Plotly,
    title=f'Parallel Coordinates Plot for K-Means Clustering (k={n_clusters})'
)

# Customize the plot
fig.update_layout(
    plot_bgcolor='white',
    paper_bgcolor='white',
    font=dict(size=10),
    margin=dict(l=50, r=50, b=50, t=50)
)

# Show the plot
fig.show()
```
