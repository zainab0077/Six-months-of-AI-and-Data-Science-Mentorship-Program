# Kurtosis and Skewness
Close to zero means data is normalized 

```python
# Lets explore for the passengers only 
from scipy import stats
print(f"Kurtosis of Fare : {stats.kurtosis(df_01['fare'],fisher=False)}")
print(f"Skewness of Fare : {stats.skew(df_01['fare'])}")
px.histogram(df_01,x='fare',color='who')
```

```python
# Lets explore for the second class passengers only 
df_second = df_01[df_01['class']=='Second']
print(f"Kurtosis of Fare : {stats.kurtosis(df_second['fare'],fisher=False)}")
print(f"Skewness of Fare : {stats.skew(df_second['fare'])}")
px.histogram(df_second,x='fare',color='who')
```

```python
# Lets explore for the third class passengers only 
df_third = df_01[df_01['class']=='Third']
print(f"Kurtosis of Fare : {stats.kurtosis(df_third['fare'],fisher=False)}")
print(f"Skewness of Fare : {stats.skew(df_third['fare'])}")
px.histogram(df_third,x='fare',color='who')
```

# Data Normality Test 

```python
stats.shapiro(df_01['fare'])
```

```python
# shapiro wilk test to check normalization
print(f"Test Statistics : {stats.shapiro(df_01['fare'])[0]:.4f}")
print(f"P Value : {stats.shapiro(df_01['fare'])[1]:.4f}")
```
