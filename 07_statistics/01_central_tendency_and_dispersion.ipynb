## 2.IQR

```python
np.random.seed(0)
data = np.random.normal(100,20,200)
data = np.append(data,[300,5])

# find q1 and q3
q1 = np.percentile(data,25)
q3 = np.percentile(data,75)
IQR = q3-q1

# find lower and upper bound
lowerbound = q1-1.5*IQR
upperbound = q3+1.5*IQR

# identify outliers 
outlier = (data<lowerbound) | (data>upperbound)
# filter clean data
data_cleaned = data[~outlier]
# plot the data 
plt.figure(figsize = (14,6))
plt.subplot(1,2,1)
plt.boxplot(data)
plt.title('Original Data with outliers')

# plot the cleaned data 
plt.subplot(1,2,2)
plt.boxplot(data_cleaned)
plt.title('Data After Removing Outliers')

plt.show()
```

## 3.Variance

```python
df = np.array([2,3,4,5,6])
mean = np.mean(df)
print(f"Mean of Data : {mean}")
variance = np.var(df,ddof=1)
print(f"Variance of Data : {variance}")
plt.bar(['Mean','Variance'],[mean,variance],color=['blue','orange'])
plt.title('Mean and Variance of the dataset')
```

## 4.Standard Deviation

```python
data = np.array([2,3,4,5,6])
std = np.std(data,ddof=1)
print(f"Standard Deviation of Data : {std:.4f}")
```
